## v0. Base Hangman Logic

We’ll start with the simplest version: a game that picks a random word and lets you guess letters.

```python
import random

words = ['python', 'hangman', 'challenge', 'programming', 'adventure']

def base_hangman(guess, state={}):
    # Initialize game state
    if 'word' not in state:
        state['word'] = random.choice(words)
        state['guessed'] = set()
        state['game_over'] = False

    if state['game_over']:
        return f"Game over! The word was '{state['word']}'", state

    # Validate the input
    guess = guess.lower()
    if len(guess) != 1 or not guess.isalpha():
        return "Please enter a single letter.", state

    if guess in state['guessed']:
        return f"You already guessed '{guess}'.", state

    state['guessed'].add(guess)

    # Display current progress
    display = ' '.join([c if c in state['guessed'] else '_' for c in state['word']])

    if display.replace(' ', '') == state['word']:
        state['game_over'] = True
        return f"You won! The word was '{state['word']}' 🎉", state

    return f"Word: {display}", state
```

Let's pilot run the text-based game!

```python
# Start the game
while True:
    display_str, state = base_hangman(input("Enter a letter: "))
    print(display_str)
    if state['game_over']:
        break
```

### Feature 2: Serving Hangman with Gradio

Cool, now our game is the right difficulty.
But we want our friends to be able to play our game.
Let’s make it playable in the browser with Gradio.

#### ☝️🤓 What's Gradio?
Gradio is a Python library that makes it easy to turn your code into a simple web app with a user interface.

Instead of running programs only in the console, you can add text boxes, buttons, and sliders, and then share a link so others can try your project directly in their browser — without installing anything.

In short: **Gradio = a quick way to make your Python programs interactive and shareable on the web.**

```python
import gradio as gr

def on_submit(guess, state):
    text, state = hangman_with_wrong_count(guess, state)
    return "", text, state  # Clear input, update output, keep state

with gr.Blocks() as hangman_app:
    state = gr.State({})
    gr.Markdown("## 🕹️ Hangman Game (v1)\nEnter a letter to guess the word!")

    guess_input = gr.Textbox(label="Your guess (single letter)")
    output = gr.Textbox(label="Game Output", interactive=False, lines=10)

    guess_input.submit(on_submit, [guess_input, state], [guess_input, output, state])

# Run the app
hangman_app.launch(share=True)
```
