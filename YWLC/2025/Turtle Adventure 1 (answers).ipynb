## 8. Setting up the Map

### 📚 What are libraries?

You’ve already learned how to use variables, math, and print statements — now let’s make things move!

When we use Python, we start with a basic set of tools — things like numbers, strings, loops, and `print()`.

But sometimes we want to use extra tools that other people have made, like:

* Drawing turtles 🐢

* Doing math and graphs 📊

* Building games 🎮

These extra toolkits are called libraries or packages.

### 🐢 What is ColabTurtlePlus?
*ColabTurtlePlus* is a small Python library that lets you draw and move a little turtle on the screen, kind of like a mini robot.
You can tell the turtle things like:

* “move forward 50 steps”

* “turn right 90 degrees”

* “go faster” or “go slower”

And the turtle will actually do it — you’ll see it move!

### 🧩 How do we get them?

We use a helper program called pip.
pip is like an app store for Python — it can download and install libraries from the internet so you can use them in your code.

To install a library, you type this in your notebook or terminal:

`!pip install ColabTurtlePlus`


The `!` at the start just tells Google Colab: “run this as a command, not as Python code.”

* `pip install` means: “download and set up this library.”

* `ColabTurtlePlus` is the name of the library we want.

You only need to install a library once in your environment.
After that, you can just import it in your Python code whenever you need it.

````python
!pip install ColabTurtlePlus
````

### 🧩 Why do we need to “import” it?

When you install Python, it comes with the most common tools already built in — like numbers, strings, and print().
But ColabTurtlePlus is a special tool made by someone else.

To use it, we have to import it — that just means:

“Hey Python, please bring in this extra toolbox so I can use its commands.”

For example:

`from ColabTurtlePlus.Turtle import *`


This line tells Python:

“Get all the turtle commands (like `forward()`, `left()`, `speed()`) from the ColabTurtlePlus library.”

🐢 Then we can do fun stuff like:
```python
initializeTurtle() # Create turtle
speed(10)          # Set turtle's speed
forward(100)       # Move forward
right(90)          # Turn right
forward(50)        # Move again
```

Now you’ve got a moving turtle drawing lines on your screen!

## 11. The Turtle Learns Geometry 📐

````python
reset()
sides = 6
length = 70

for i in range(sides):
    forward(length)
    right(360 / sides)
````

## 12. The Turtle’s Magic Spell ✨

````python
reset()
def draw_shape(sides, length):
    for i in range(sides):
        forward(length)
        right(360 / sides)

draw_shape(8, 40)  # Turtle draws an octagon spell
````

## 13. Turtle Fireworks 🎆

````python
reset()
speed(10)

for i in range(50):
    forward(i * 5)
    right(45)
````
